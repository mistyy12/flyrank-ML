## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

This playbook is intended for content editors and page owners who need to prioritize pages for review. The logistic regression model ranks pages by predicted decline risk, while the action rules suggest what a reviewer should investigate.

The model predicts a decline label; it does not measure content quality or prove that refreshing a page will improve performance. Its findings are limited to the clients, eligible pages, time windows, and validation design evaluated. Performance on other clients or future periods may differ. Pages with insufficient usable history require a data-quality check before recommendations can be trusted.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

Before acting, a reviewer should:

- Confirm that tracking is working and the page has sufficient usable history.

- Check recent performance and consider seasonality, technical issues, or changes in search demand.

- Inspect the page for outdated information, factual errors, and mismatch with search intent.

- Compare the suggested action with the page’s business relevance and estimated editing effort.

- Approve, modify, or reject the recommendation and record the reason.

The model must not automatically publish edits, delete or prune pages, redirect URLs, or change indexing settings. Low CTR or a high decline score alone is insufficient evidence for these actions. Sensitive or specialist content requires an appropriately qualified reviewer.

If evidence is incomplete or conflicting, the page should be investigated or monitored before any content change is approved.
