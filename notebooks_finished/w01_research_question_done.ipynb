```python
#1 Stale Pages
stale_pages = (df["days_since_last_update"]>=180).sum()
print(stale_pages)


#2. Older content with meaningful search demand
hi_demand = ((df["days_since_last_update"] >= 90) & (df["impressions_90d"] >= 500)).sum()
print(hi_demand)

#3. Lower Recent Clicks
df["expected_clicks_30d"] = df["clicks_90d"]/3
df["lower_clicks"] = (df["clicks_prev_30d"] < df["expected_clicks_30d"])
lower_clicks = df["lower_clicks"].sum()
print(lower_clicks)
```
